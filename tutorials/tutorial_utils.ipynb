## Search

There are two main ways to search for a clock in `pyaging`. The first is through the doi of the paper in which the clock was developed.

```python
pya.utils.find_clock_by_doi('https://doi.org/10.1038/s43587-022-00248-2')
```

The second way is by simply showing the names of all the clocks that are available.

```python
pya.utils.show_all_clocks()
```

## Feature units and normal ranges

Clinical clocks are only as good as the units they are fed, so `pyaging` keeps a registry of the unit and the plausible range of every feature it knows about. The registry is package-wide ground truth: there is one entry per feature name, shared by every clock that uses it, so `albumin` means g/L whether you are running PhenoAge, KDM-Age, or LinAge2. To see what a particular clock expects:

```python
pya.utils.get_feature_ranges('PhenoAge')
```

`low` and `high` are plausibility bounds rather than reference intervals — wide enough to admit a genuinely sick person, narrow enough to catch a decimal point or a swapped column. At prediction time `pyaging` compares your inputs against them and **warns**; it never raises. An out-of-range value is usually a unit mistake, but it can be real, and refusing to predict would be the wrong call.

Features the registry does not list fall back to a default for the clock's modality. Every DNA methylation clock, for instance, inherits the same beta-value bounds without listing hundreds of thousands of CpGs individually:

```python
pya.utils.get_feature_ranges('Horvath2013').head()
```

Units are not always physical. LinAge2 takes 26 questionnaire items straight from NHANES, and the registry records their answer codes so you can see what a `1` or a `2` is supposed to mean:

```python
pya.utils.get_feature_ranges('LinAge2').tail(6)
```
