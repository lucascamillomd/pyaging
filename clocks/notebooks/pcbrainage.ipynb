# PCBrainAge

Built from the original authors’ pinned code and coefficients. See `tests/data/pc_extensions/README.md` for source checksums, exact linear composition, and independent R oracle validation.


```python
from pathlib import Path
import sys
import torch

root = Path.cwd().resolve()
while not (root / "clocks/build_pc_extensions.py").exists():
    if root == root.parent:
        raise RuntimeError("Run this notebook inside the pyaging repository")
    root = root.parent
sys.path.insert(0, str(root / "clocks"))
from build_pc_extensions import build_model

model = build_model('pcbrainage')
```

```python
model.metadata["clock_name"] = 'pcbrainage'
model.metadata["data_type"] = 'DNA methylation'  # Paper: The predictor uses DNA methylation from postmortem dorsolateral prefrontal cortex.
model.metadata["species"] = 'Homo sapiens'  # Paper: The training dataset comprises human dorsolateral prefrontal cortex samples.
model.metadata["year"] = 2022
model.metadata["approved_by_author"] = '⌛'
model.metadata["citation"] = 'Thrush, K. L., Bennett, D. A., Gaiteri, C., Horvath, S., van Dyck, C. H., Higgins-Chen, A. T., and Levine, M. E. (2022). Aging the brain: multi-region methylation principal component based clock in the context of Alzheimer’s disease. Aging 14(14), 5641–5668.'
model.metadata["doi"] = 'https://doi.org/10.18632/aging.204196'
model.metadata["notes"] = 'Brain chronological-age clock trained in dorsolateral prefrontal cortex. The original PCA and regression are composed algebraically without retraining; missing CpGs use the authors’ GSE74193 adult brain means. No age-acceleration residualization is performed.'
model.metadata["research_only"] = None
model.metadata["tissue"] = ['brain cortex']  # Paper: Training used dorsolateral prefrontal cortex from GSE74193.
model.metadata["predicts"] = ['chronological age']  # Paper: The model outcome variable is chronological age.
model.metadata["training_target"] = ['chronological age']  # Paper: The model outcome variable is chronological age.
model.metadata["unit"] = ['years']  # Paper: The predictor estimates chronological age in years.
model.metadata["model_type"] = 'PCA + elastic net regression'  # Paper: Principal components were supplied to elastic-net penalized regression.
model.metadata["platform"] = ['Illumina 450K']  # Paper: GSE74193 methylation was measured using the Illumina 450K array.
model.metadata["population"] = 'adults'  # Paper: The training dataset contains 399 individuals aged 20 years and older.
model.metadata["journal"] = 'Aging'
model.metadata["last_author"] = 'Morgan E. Levine'
model.metadata["n_features"] = 357852
model.metadata["citations"] = 40
model.metadata["citations_date"] = '2026-10-02'
```

```python
model.eval()
model.to(torch.float64)
output = root / "clocks/weights" / (model.metadata["clock_name"] + ".pt")
output.parent.mkdir(parents=True, exist_ok=True)
torch.save(model, output)
output
```
