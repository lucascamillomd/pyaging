Let's first import some packages:

```python
import os
import inspect
import shutil
import json
import math
import torch
import pandas as pd
import pyaging as pya
```

## Instantiate model class

```python
def print_entire_class(cls):
    source = inspect.getsource(cls)
    print(source)

print_entire_class(pya.models.YChrom)
```

```python
model = pya.models.YChrom()
```

## Define clock metadata

```python
model.metadata["clock_name"] = "ychrom"
model.metadata["data_type"] = "DNA methylation"  # Paper: A methylation-based sex classifier uses separate X and Y principal components.
model.metadata["species"] = "Homo sapiens"  # Paper: A methylation-based sex classifier uses separate X and Y principal components.
model.metadata["year"] = 2021
model.metadata["approved_by_author"] = "⌛"
model.metadata["citation"] = "Wang, Y. et al. DNA methylation-based sex classifier to predict sex and identify sex chromosome aneuploidy. BMC Genomics 22, 484 (2021)."
model.metadata["doi"] = "https://doi.org/10.1186/s12864-021-07675-2"
model.metadata["notes"] = "Y-chromosome first-principal-component score used with the X score to infer Y presence and classify 46,XX, 46,XY, 45,X, and 47,XXY samples."
model.metadata["research_only"] = None
model.metadata["tissue"] = ["whole blood"]  # Paper: The classifier was trained and validated using whole blood.
model.metadata["predicts"] = ["Y-chromosome presence"]  # Paper: Y PC1 separates samples with versus without a Y chromosome.
model.metadata["training_target"] = ["Y-chromosome presence"]  # Paper: PCA maximized variance after sex-associated probe screening.
model.metadata["unit"] = ["unitless"]  # Paper: The first principal component is used as the chromosome score.
model.metadata["model_type"] = "principal component analysis"  # Paper: Separate PCA models were fitted to X- and Y-chromosome probes.
model.metadata["platform"] = ["Illumina 450K", "Illumina EPIC"]  # Paper: The classifier is compatible with 450K and EPIC arrays.
model.metadata["population"] = "adults"  # Paper: 800 females and 800 males from age-18 GSE105018 were selected.
model.metadata["journal"] = "BMC Genomics"
model.metadata["last_author"] = "Leonard C. Schalkwyk"
model.metadata["n_features"] = 284
model.metadata["citations"] = 38
model.metadata["citations_date"] = "2026-07-05"
```

## Download clock dependencies

#### Download coefficient file

```python
coeff_url = "https://raw.githubusercontent.com/bio-learn/biolearn/master/biolearn/data/estimateSex.csv"
os.system(f"curl -L {coeff_url} -o estimateSex.csv")
```

## Load features

#### From CSV file

```python
import numpy as np
coeffs = pd.read_csv('estimateSex.csv')
coeffs['feature'] = coeffs['IlmnID']
model.features = coeffs['feature'].tolist()
```

## Load weights into base model

#### From CSV file

```python
x_mask = coeffs['pca'] == 'X'
y_mask = coeffs['pca'] == 'Y'
autosome_mask = ~coeffs['CHR'].isin(['X', 'Y'])

model.autosome_indices = torch.tensor(np.where(autosome_mask)[0], dtype=torch.long)
if 'Y' == 'X':
    model.x_indices = torch.tensor(np.where(x_mask)[0], dtype=torch.long)
    model.x_means = torch.tensor(coeffs.loc[x_mask, 'mean'].values, dtype=torch.float32)
    model.x_coeffs = torch.tensor(coeffs.loc[x_mask, 'coeff'].values, dtype=torch.float32)
else:
    model.y_indices = torch.tensor(np.where(y_mask)[0], dtype=torch.long)
    model.y_means = torch.tensor(coeffs.loc[y_mask, 'mean'].values, dtype=torch.float32)
    model.y_coeffs = torch.tensor(coeffs.loc[y_mask, 'coeff'].values, dtype=torch.float32)

model.reference_values = [float("nan")] * len(model.features)
```

# No base model needed; scores computed in forward


```python
model.base_model = None
```

## Load preprocess and postprocess objects

```python
model.preprocess_name = "sex_estimation_autosomal_zscore"
model.preprocess_dependencies = [model.autosome_indices.tolist()]
```

```python
model.postprocess_name = None
model.postprocess_dependencies = None
```

## Check all clock parameters

```python
pya.utils.print_model_details(model)
```

## Normal feature ranges

```python
# Units and plausibility ranges come from the package registry, keyed by feature name.
feature_ranges = pya.utils.resolve_feature_ranges(model.features, model.metadata["data_type"])
model.feature_units = [record["unit"] for record in feature_ranges]
pd.DataFrame.from_records(feature_ranges).head()
```

## Basic Test

```python
# Exercise the clock with values in the middle of each feature's expected range.
records = pya.utils.resolve_feature_ranges(model.features, model.metadata["data_type"])
midpoints = [
    (record["low"] + record["high"]) / 2 if math.isfinite(record["high"]) else max(record["low"], 1.0)
    for record in records
]
input = torch.tensor([midpoints] * 10, dtype=torch.float64)
model.eval()
model.to(torch.float64)
pred = model(input)
pred
```

## Save torch model


```python
torch.save(model, f"../weights/{model.metadata['clock_name']}.pt")
```

```python
# Function to remove a folder and all its contents
def remove_folder(path):
    try:
        shutil.rmtree(path)
        print(f"Deleted folder: {path}")
    except Exception as e:
        print(f"Error deleting folder {path}: {e}")

# Get a list of all files and folders in the current directory
all_items = os.listdir('.')

# Loop through the items
for item in all_items:
    # Check if it's a file and does not end with .ipynb
    if os.path.isfile(item) and not item.endswith('.ipynb'):
        os.remove(item)
        print(f"Deleted file: {item}")
    # Check if it's a folder
    elif os.path.isdir(item):
        remove_folder(item)
```
