# PCGrimAgeB2M

Built from the original authors’ pinned code and coefficients. See `tests/data/pc_extensions/README.md` for source checksums, exact linear composition, and independent R oracle validation.


```python
from pathlib import Path
import sys
import torch

root = Path.cwd().resolve()
while not (root / "clocks/build_pc_extensions.py").exists():
    if root == root.parent:
        raise RuntimeError("Run this notebook inside the pyaging repository")
    root = root.parent
sys.path.insert(0, str(root / "clocks"))
from build_pc_extensions import build_model

model = build_model('pcgrimageb2m')
```

```python
model.metadata["clock_name"] = 'pcgrimageb2m'
model.metadata["data_type"] = 'DNA methylation'  # Paper: The study constructs clocks from DNA methylation measurements.
model.metadata["species"] = 'Homo sapiens'  # Paper: The analyzed samples and clock are human.
model.metadata["year"] = 2022
model.metadata["approved_by_author"] = '⌛'
model.metadata["citation"] = 'Higgins-Chen, Albert T., et al. "A computational solution for bolstering reliability of epigenetic clocks: implications for clinical trials and longitudinal tracking." Nature Aging 2 (2022): 644–661.'
model.metadata["doi"] = 'https://doi.org/10.1038/s43587-022-00248-2'
model.metadata["notes"] = 'Standalone principal-component proxy trained to reproduce the original DNAm beta-2-microglobulin score. Coefficients are composed exactly from the author PCA rotation and proxy regression. CpGs use the author GSE40279 reference. Sex (female=1, male=0) and chronological age must be supplied when selected by the original component regression; missing demographic inputs are rejected. Selected demographic inputs: age.'
model.metadata["research_only"] = None
model.metadata["tissue"] = ['whole blood']  # Paper: The training datasets represent whole blood.
model.metadata["predicts"] = ['beta-2-microglobulin']  # Paper: The standalone PC-based component reproduces the original DNAm beta-2-microglobulin output; the released author code calculates this component before combining PCGrimAge.
model.metadata["training_target"] = ['DNAm beta-2-microglobulin output']  # Paper: The standalone PC-based component reproduces the original DNAm beta-2-microglobulin output; the released author code calculates this component before combining PCGrimAge.
model.metadata["unit"] = ['picograms per milliliter']  # Paper: The original GrimAge smoking surrogate is in pack-years and plasma protein surrogates are in picograms per milliliter.
model.metadata["model_type"] = 'PCA + elastic net regression'  # Paper: Principal components were supplied to elastic-net regression.
model.metadata["platform"] = ['Illumina 450K']  # Paper: The listed arrays are those used for the model's training datasets.
model.metadata["population"] = 'adults'  # Paper: Training cohorts and sample types are specified in the study dataset table.
model.metadata["journal"] = 'Nature Aging'
model.metadata["last_author"] = 'Morgan E. Levine'
model.metadata["n_features"] = 78465
model.metadata["citations"] = 546
model.metadata["citations_date"] = '2026-10-02'
```

```python
model.eval()
model.to(torch.float64)
output = root / "clocks/weights" / (model.metadata["clock_name"] + ".pt")
output.parent.mkdir(parents=True, exist_ok=True)
torch.save(model, output)
output
```
