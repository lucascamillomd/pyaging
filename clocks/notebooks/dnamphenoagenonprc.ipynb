# dnamphenoagenonprc
Rebuild from the original authors using `clocks/build_author_extensions.py`. Source revisions and checksums are pinned in that script.
Original Levine-lab non-polycomb PhenoAge contribution using 458 CpGs. Coefficients retain their original values; the 60.664 PhenoAge intercept is excluded from both partitions, so these scores are not standalone calibrated ages. The two contributions plus the original intercept reconstruct DNAmPhenoAge. Follows author imputation=FALSE: absent CpGs and supplied NA values contribute zero.
Build-only dependencies: `rdata` and `remotezip`; run the builder with `uv run --with rdata --with remotezip`. No R dependency is needed for Python prediction.


```python
from pathlib import Path
import sys
import torch
root = Path.cwd()
while not (root / "pyproject.toml").exists():
    root = root.parent
sys.path.insert(0, str(root / "clocks"))
from build_author_extensions import sources, build_models
intrin, prc = sources(Path("/tmp/pyaging-author-extensions"))
model = build_models(intrin, prc)['dnamphenoagenonprc']
```

```python
model.metadata['clock_name'] = 'dnamphenoagenonprc'
model.metadata['data_type'] = 'DNA methylation'  # Paper: methylation
model.metadata['species'] = 'Homo sapiens'  # Paper: Homo sapiens
model.metadata['year'] = 2022
model.metadata['approved_by_author'] = '⌛'
model.metadata['citation'] = ['Rozenblit, M., et al. Evidence of accelerated epigenetic aging of breast tissues in patients with breast cancer is driven by CpGs associated with polycomb-related genes. Clinical Epigenetics 14, 30 (2022).', 'Levine, M. E., et al. "An epigenetic biomarker of aging for lifespan and healthspan." Aging 10.4 (2018): 573-591.']
model.metadata['doi'] = 'https://doi.org/10.1186/s13148-022-01249-z'
model.metadata['notes'] = 'Original Levine-lab non-polycomb PhenoAge contribution using 458 CpGs. Coefficients retain their original values; the 60.664 PhenoAge intercept is excluded from both partitions, so these scores are not standalone calibrated ages. The two contributions plus the original intercept reconstruct DNAmPhenoAge. Follows author imputation=FALSE: absent CpGs and supplied NA values contribute zero. The parent DNAmPhenoAge coefficients were fitted on InCHIANTI 450K whole-blood data; the selected sites also occur on 27K and EPIC arrays.'
model.metadata['research_only'] = None
model.metadata['tissue'] = ['whole blood']  # Paper: whole blood
model.metadata['predicts'] = ['non-polycomb PhenoAge contribution']  # Paper: The author package returns the non-polycomb PhenoAge subscore.
model.metadata['training_target'] = ['phenotypic age']  # Paper: phenotypic age
model.metadata['unit'] = ['years']  # Paper: The returned contribution retains the PhenoAge coefficient scale; no age intercept is added.
model.metadata['model_type'] = 'weighted linear score'  # Paper: Each subscore is a weighted sum of the original selected CpGs.
model.metadata['platform'] = ['Illumina 450K']  # Paper: The parent DNAmPhenoAge model was fitted in InCHIANTI using NOOB-normalized 450K whole-blood methylation. Common 27K/450K/EPIC sites were chosen for compatibility; JHS EPIC data were validation.
model.metadata['population'] = 'adults'  # Paper: adults (21-100 years)
model.metadata['journal'] = 'Clinical Epigenetics'
model.metadata['last_author'] = 'Morgan Levine'
model.metadata['n_features'] = 458
model.metadata['citations'] = 23
model.metadata['citations_date'] = '2026-10-02'
```

```python
model.version = "0.5.7"
model.metadata["version"] = "0.5.7"
(root / "clocks/weights").mkdir(parents=True, exist_ok=True)
torch.save(model, root / "clocks/weights/dnamphenoagenonprc.pt")
```
