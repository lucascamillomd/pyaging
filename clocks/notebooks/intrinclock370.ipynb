# intrinclock370
Rebuild from the original authors using `clocks/build_author_extensions.py`. Source revisions and checksums are pinned in that script.
Author-demo IntrinClock at lambda.1se: 370 nonzero CpGs from the original serialized cv.glmnet fit. Predicts chronological age resistant to immune-cell-composition changes. This is a separately fitted regularization column, not the existing 380-CpG lambda.min model with ten sites removed. Returns years using the author's inverse Horvath transform.
Build-only dependencies: `rdata` and `remotezip`; run the builder with `uv run --with rdata --with remotezip`. No R dependency is needed for Python prediction.


```python
from pathlib import Path
import sys
import torch
root = Path.cwd()
while not (root / "pyproject.toml").exists():
    root = root.parent
sys.path.insert(0, str(root / "clocks"))
from build_author_extensions import sources, build_models
intrin, prc = sources(Path("/tmp/pyaging-author-extensions"))
model = build_models(intrin, prc)['intrinclock370']
```

```python
model.metadata['clock_name'] = 'intrinclock370'
model.metadata['data_type'] = 'DNA methylation'  # Paper: IntrinClock uses CpG DNA methylation beta values.
model.metadata['species'] = 'Homo sapiens'  # Paper: The clock database and sorted immune-cell experiments used human samples.
model.metadata['year'] = 2024
model.metadata['approved_by_author'] = '⌛'
model.metadata['citation'] = 'Tomusiak, A., et al. “Development of an epigenetic clock resistant to changes in immune cell composition.” Communications Biology 7, 934 (2024).'
model.metadata['doi'] = 'https://doi.org/10.1038/s42003-024-06609-4'
model.metadata['notes'] = "Author-demo IntrinClock at lambda.1se: 370 nonzero CpGs from the original serialized cv.glmnet fit. Predicts chronological age resistant to immune-cell-composition changes. This is a separately fitted regularization column, not the existing 380-CpG lambda.min model with ten sites removed. Returns years using the author's inverse Horvath transform."
model.metadata['research_only'] = None
model.metadata['tissue'] = ['multi-tissue']  # Paper: Training used samples from many tissues, with the majority from blood.
model.metadata['predicts'] = ['chronological age']  # Paper: IntrinClock was developed as an epigenetic age predictor and evaluated against chronological age.
model.metadata['training_target'] = ['chronological age']  # Paper: Ages were transformed using Horvath's original clock formula before training.
model.metadata['unit'] = ['years']  # Paper: returnAge uses 21*exp(score)-1 below zero and 21*score+20 otherwise.
model.metadata['model_type'] = 'two-stage elastic net regression'  # Paper: Two elastic-net glmnet models were fit with alpha 0.5 and tenfold cross-validation.
model.metadata['platform'] = ['Illumina 450K', 'Illumina EPIC']  # Paper: The fitting database contained samples measured on 450K or EPIC arrays.
model.metadata['population'] = 'all ages'  # Paper: After quality control and a 75:25 split, 9,104 samples trained the model and 2,994 validated it.
model.metadata['journal'] = 'Communications Biology'
model.metadata['last_author'] = 'Eric Verdin'
model.metadata['n_features'] = 370
model.metadata['citations'] = 64
model.metadata['citations_date'] = '2026-10-02'
```

```python
model.version = "0.5.4"
model.metadata["version"] = "0.5.4"
(root / "clocks/weights").mkdir(parents=True, exist_ok=True)
torch.save(model, root / "clocks/weights/intrinclock370.pt")
```
