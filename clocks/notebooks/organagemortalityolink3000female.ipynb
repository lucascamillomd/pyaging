Source: [organAging commit 5147b0301ec7f4abdb10ef650d04f47454ddc8fd](https://github.com/ludgergoeminne/organAging/tree/5147b0301ec7f4abdb10ef650d04f47454ddc8fd). Download hashes and conversion logic are in `clocks/build_organage.py`. **Academic/non-commercial research only**: the author license is embedded in the model and written beside the weight file.

Input: case-sensitive original protein symbols with Olink NPX values. Absent proteins have zero contribution; explicit NaNs propagate. Optional scaling for non-Olink data must be performed before prediction using the authors' guidance. Output: relative natural-log mortality hazard.

```python
from pathlib import Path
import sys

root = next(path for path in (Path.cwd(), *Path.cwd().parents) if (path / "clocks" / "build_organage.py").is_file())
sys.path.insert(0, str(root / "clocks"))
from build_organage import build_clock, save_clock

model = build_clock('organagemortalityolink3000female')
```

```python
model.metadata["clock_name"] = 'organagemortalityolink3000female'
model.metadata["data_type"] = 'proteomics'  # Paper: Plasma protein-based organ-specific aging and mortality models.
model.metadata["species"] = 'Homo sapiens'  # Paper: Models were developed using human UK Biobank plasma proteomics.
model.metadata["year"] = 2025
model.metadata["approved_by_author"] = '⌛'
model.metadata["citation"] = 'Goeminne, L. J. E., et al. "Plasma protein-based organ-specific aging and mortality models unveil diseases as accelerated aging of organismal systems." Cell Metabolism 37, 205–222.e6 (2025).'
model.metadata["doi"] = 'https://doi.org/10.1016/j.cmet.2024.10.005'
model.metadata["notes"] = 'Female OrganAge mortality model for Olink Explore 3072, using author-recommended fold 1. Input columns are the original case-sensitive protein symbols and values are Olink NPX, already normalized. No additional centering or cohort scaling is applied. Absent proteins contribute zero; NaNs in supplied proteins propagate, matching the original Example_Script.R scoring expressions. For other assays, prepare and rescale protein values according to the author guidance before prediction. Output is relative natural-log mortality hazard, not age in years. Models labelled Female and Male contain protein features; neither requires a sex covariate. Academic/non-commercial research use only; commercial use requires a separate license from the authors. This additional author-released model was not discussed in the main manuscript.'
model.metadata["research_only"] = True
model.metadata["tissue"] = ['blood']  # Paper: Plasma protein-based organ-specific aging and mortality models.
model.metadata["predicts"] = ['mortality risk']  # Paper: The result will be the predicted biological age or the predicted relative log(mortality hazard).
model.metadata["training_target"] = ['mortality']  # Paper: Mortality-based models were trained to predict time to death.
model.metadata["unit"] = ['log hazard']  # Paper: Mortality-based models generate relative log(mortality hazards).
model.metadata["model_type"] = 'elastic net Cox regression'  # Paper: The mortality-based models were trained with Cox elastic net.
model.metadata["platform"] = ['Olink Explore 3072']  # Paper: The authors provide dedicated coefficients for Olink Explore 3072 data.
model.metadata["population"] = 'adults'  # Paper: UK Biobank participants were recruited as adults.
model.metadata["journal"] = 'Cell Metabolism'
model.metadata["last_author"] = 'Vadim N. Gladyshev'
model.metadata["n_features"] = 6
model.metadata["citations"] = 128
model.metadata["citations_date"] = '2026-10-02'
```

```python
save_clock(model)
print(model.metadata["clock_name"], len(model.features), model.metadata["unit"])
```
