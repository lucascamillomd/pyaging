# HPS

Healthspan Proteomic Score (HPS) from the original Kuo lab R release: 86 Olink Explore 3072 plasma protein NPX values plus chronological age in years. Returns the modeled probability of remaining free of the study's major-disease-or-death endpoint over the next 10 years, on a 0-1 scale; higher is healthier. It is not predicted years of life, biological age, or an age-adjusted residual. The endpoint includes cancer excluding nonmelanoma skin cancer, diabetes, heart failure, myocardial infarction, stroke, COPD, dementia, or death. Supply original lowercase identifiers, including separate nppb and ntprobnp assays. NPX is assay-normalized log2 relative abundance, not raw counts, concentrations, linear abundances or z-scores. The study used within-batch and across-batch intensity normalization; NPX labeling alone does not establish comparability across panels, batches, plasma and serum. No additional scaling, normalization or imputation occurs during scoring. All 87 predictor columns are required; supplied NaNs propagate. The study used upstream k-nearest-neighbor imputation (k=10, multiUS); any imputation is a separate caller decision. The Gompertz model was fit in 30,184 UK Biobank adults initially free of the specified endpoint conditions; the healthy cohort age range was 39-70. The August 2025 correction only changes author affiliations. Research use only. The original repository specifies no code license; the publication is CC BY-NC-ND 4.0. This independent Torch implementation uses the published fitted numerical parameters and does not relicense the authors' R code. No author approval of this conversion is claimed.

[Original publication](https://doi.org/10.1073/pnas.2414086122) and [pinned author R source](https://github.com/kuo-lab-uchc/HPS/blob/6eec9ac5a091e34fece69fa4652ebd9af22c1839/HPS.R). The builder verifies SHA-256 digests before extracting the fitted numerical parameters. Native R reference checks are documented in `tests/data/hps`.

Supply a samples-by-features matrix with the original lowercase NPX protein names and `age` in years. Preserve both `nppb` and `ntprobnp`, which are separate model inputs. Do not place `age` in `metadata_cols`. Construct `AnnData` directly to preserve supplied NaNs; any upstream imputation is a separate choice.
