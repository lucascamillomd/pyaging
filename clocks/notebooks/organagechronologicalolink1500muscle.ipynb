Input: case-sensitive original protein symbols with Olink NPX values. Absent proteins have zero contribution; explicit NaNs propagate. Optional scaling for non-Olink data must be performed before prediction using the authors' guidance. Output: predicted age in years.

```python
from pathlib import Path
import sys

root = next(path for path in (Path.cwd(), *Path.cwd().parents) if (path / "clocks" / "build_organage.py").is_file())
sys.path.insert(0, str(root / "clocks"))
from build_organage import build_clock, save_clock

model = build_clock('organagechronologicalolink1500muscle')
```

```python
model.metadata["clock_name"] = 'organagechronologicalolink1500muscle'
model.metadata["data_type"] = 'proteomics'  # Paper: Plasma protein-based organ-specific aging and mortality models.
model.metadata["species"] = 'Homo sapiens'  # Paper: Models were developed using human UK Biobank plasma proteomics.
model.metadata["year"] = 2025
model.metadata["approved_by_author"] = '⌛'
model.metadata["citation"] = 'Goeminne, L. J. E., et al. "Plasma protein-based organ-specific aging and mortality models unveil diseases as accelerated aging of organismal systems." Cell Metabolism 37, 205–222.e6 (2025).'
model.metadata["doi"] = 'https://doi.org/10.1016/j.cmet.2024.10.005'
model.metadata["notes"] = 'Muscle OrganAge chronological model for Olink Explore 1536, using author-recommended fold 1. Input columns are the original case-sensitive protein symbols and values are Olink NPX, already normalized. No additional centering or cohort scaling is applied. Absent proteins contribute zero; NaNs in supplied proteins propagate, matching the original Example_Script.R scoring expressions. For other assays, prepare and rescale protein values according to the author guidance before prediction. Output is predicted age in years. Models labelled Female and Male contain protein features; neither requires a sex covariate. Academic/non-commercial research use only; commercial use requires a separate license from the authors.'
model.metadata["research_only"] = True
model.metadata["tissue"] = ['blood']  # Paper: Plasma protein-based organ-specific aging and mortality models.
model.metadata["predicts"] = ['biological age']  # Paper: The result will be the predicted biological age or the predicted relative log(mortality hazard).
model.metadata["training_target"] = ['chronological age']  # Paper: First-generation models were trained to predict chronological age.
model.metadata["unit"] = ['years']  # Paper: First-generation models predict age in years.
model.metadata["model_type"] = 'elastic net regression'  # Paper: First-generation models were trained using elastic net regression.
model.metadata["platform"] = ['Olink Explore 1536']  # Paper: The authors provide dedicated coefficients for Olink Explore 1536 data.
model.metadata["population"] = 'adults'  # Paper: UK Biobank participants were recruited as adults.
model.metadata["journal"] = 'Cell Metabolism'
model.metadata["last_author"] = 'Vadim N. Gladyshev'
model.metadata["n_features"] = 4
model.metadata["citations"] = 128
model.metadata["citations_date"] = '2026-10-02'
```
